## Setup


```python
import numpy as np

np.set_printoptions(precision=2, suppress=True)
print("NumPy version:", np.__version__)
```

## Mini-Lesson 1: Arrays Have Shape

A NumPy array stores values in a structured numerical container. Shape tells you the dimensions.


```python
scores = np.array([82, 91, 77, 88, 95])
print(scores)
print("shape:", scores.shape)
print("dtype:", scores.dtype)
```

### Checkpoint 1: Create and Inspect an Array

Create an array of seven practice scores. Print the array, shape, mean, minimum, and maximum.


```python
practice_scores = np.array([90, 84, 76, 88, 92, 79, 85])
print(practice_scores)
print("shape:", practice_scores.shape)
print("mean:", practice_scores.mean())
print("min:", practice_scores.min())
print("max:", practice_scores.max())
```

## Mini-Lesson 2: Vectorized Operations Apply at Once

NumPy lets you transform every value without writing an explicit loop.


```python
curved_scores = scores + 5
percent_scores = scores / 100
print(curved_scores)
print(percent_scores)
```

### Checkpoint 2: Normalize Scores

Convert your practice scores from 0-100 points to 0-1 scale. Then add a 3-point curve and cap values at 100 using `np.minimum`.


```python
normalized = practice_scores / 100
curved = np.minimum(practice_scores + 3, 100)
print("normalized:", normalized)
print("curved:", curved)
```

## Mini-Lesson 3: Boolean Masks Filter Values

A Boolean mask is an array of `True` and `False` values. It selects entries that meet a condition.


```python
mask = scores >= 85
print(mask)
print(scores[mask])
```

### Checkpoint 3: Find Scores Needing Review

Use a Boolean mask to select scores below 80. Print the scores and count how many there are.


```python
needs_review = practice_scores < 80
print(practice_scores[needs_review])
print("count:", needs_review.sum())
```

### Explain the Mask

In one sentence, explain why `needs_review.sum()` counts the number of `True` values.


In Python, Boolean values are treated numerically where True equals 1 and False equals 0, so calling .sum() on a Boolean mask adds up all the 1s and effectively counts the number of True entries.


## Mini-Lesson 4: 2D Arrays Represent Tables

Rows and columns let us represent repeated observations with multiple measurements.


```python
weekly_scores = np.array([
    [82, 88, 91],
    [76, 85, 80],
    [90, 92, 95],
    [68, 74, 79],
])
print(weekly_scores)
print("shape:", weekly_scores.shape)
```

### Checkpoint 4: Row and Column Summaries

Compute each student's average across three activities and each activity's average across students.


```python
student_averages = weekly_scores.mean(axis=1)
activity_averages = weekly_scores.mean(axis=0)
print("student averages:", student_averages)
print("activity averages:", activity_averages)
```

## Mini-Lesson 5: Axis Meaning

For a 2D array, `axis=0` summarizes down rows for each column. `axis=1` summarizes across columns for each row.


```python
print("axis=0 sum:", weekly_scores.sum(axis=0))
print("axis=1 sum:", weekly_scores.sum(axis=1))
```

### Checkpoint 5: Passing Rows

Create a Boolean array showing which students have an average of at least 80. Then print only those averages.


```python

passing_student = student_averages <= 81
print(passing_student)
print(student_averages[passing_student])
passing_students = student_averages >= 80
print(passing_students)
print(student_averages[passing_students])
```

## Mini-Lesson 6: Reshape When Structure Changes

`reshape()` changes the view of the data when the total number of values stays the same.


```python
values = np.arange(1, 13)
print(values)
print(values.reshape(3, 4))
```

### Checkpoint 6: Build a 4 by 3 Array

Create the numbers 1 through 12 and reshape them into 4 rows and 3 columns. Print row sums and column sums.


```python
grid = np.arange(1, 13).reshape(4, 3)
print(grid)
print("row sums:", grid.sum(axis=1))
print("column sums:", grid.sum(axis=0))
```

### Checkpoint 7: use universal functions

Apply a square-root transformation and a cap to the adjusted score table. Keep the original and transformed arrays visible.


```python
score_table = weekly_scores
curve = np.array([2, 1, 0])
adjusted = np.minimum(score_table + curve, 100)
roots = np.sqrt(adjusted).round(2)
capped = np.minimum(adjusted * 1.05, 100)
print("roots:\n", roots)
print("capped:\n", capped)
```

**Discuss before submission:** Which operation changes the units and which operation changes the maximum allowed value?

• Changing the units: The square-root transformation (np.sqrt) changes the units of the data because it mathematically alters the scale of the original scores (e.g., a score of 81 points becomes 9.0 units).

• Changing the maximum allowed value: The np.minimum function changes the maximum allowed value by clipping or capping any score that exceeds your set limit (e.g., capping values at 100).



### Checkpoint 8: compare axis summaries

Calculate row and column totals and means after the adjustment. State which dimension each result retains.


```python
print("adjusted row totals:", adjusted.sum(axis=1))
print("adjusted column totals:", adjusted.sum(axis=0))
print("adjusted row means:", adjusted.mean(axis=1).round(1))
print("adjusted column means:", adjusted.mean(axis=0).round(1))
```

**Discuss before submission:** Why do row summaries have length three while column summaries have length four?

The prompt actually contains a common conceptual inversion: row summaries actually have a length of four, while column summaries have a length of three.

• Row summaries (axis=1) have a length of four because they calculate a single value for each of the 4 individual rows (weeks). They retain the row/time dimension.

• Column summaries (axis=0) have a length of three because they collapse the rows to calculate a single value for each of the 3 individual columns (students/activities). They retain the column/student dimension.



### Checkpoint 9: build a compound mask

Select values that are below the review threshold or above a high-score threshold, then report their positions and values.


```python
review_mask = (adjusted < 80) | (adjusted > 95)
positions = np.argwhere(review_mask)
print("mask:\n", review_mask)
print("positions:", positions.tolist())
print("values:", adjusted[review_mask])
```

**Discuss before submission:** Why is positions evidence useful in addition to the selected values?

The positions evidence is useful because the selected values array is returned as a flattened, 1D array that completely loses the structural context of the original data. While values tells you what scores triggered the thresholds, it cannot tell you who earned them or when.

The positions coordinates retain that vital context by mapping the indices directly back to the 2D grid structure, allowing you to pinpoint the exact row (week) and column (student/activity) where the threshold was breached.

### Checkpoint 10: reshape and preserve meaning

Flatten and reshape the table while recording the shape at each step. The numeric values remain but the interpretation changes when the table moves from four rows by three columns to three rows by four columns.


```python
flattened = adjusted.ravel()
reshaped = flattened.reshape(3, 4)
print("flat shape:", flattened.shape)
print("reshaped shape:", reshaped.shape)
print("reshaped:\n", reshaped)
```

**Discuss before submission:** What meaning must be re-established after changing from four rows by three columns to three rows by four columns?

When reshaping from a (4, 3) to a (3, 4) structure, you must re-establish the semantic definition of your rows and columns, because values spill over into new rows and break the original data layout (where rows meant weeks and columns meant students).
